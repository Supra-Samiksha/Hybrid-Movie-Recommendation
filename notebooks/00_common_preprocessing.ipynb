```python
import pandas as pd
import numpy as np
import os
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_PATH = PROJECT_ROOT / "dataset" / "raw"

print("Dataset path:", DATA_PATH)
print("Path exists:", DATA_PATH.exists())
```

```python
ratings = pd.read_csv(
    DATA_PATH / "user_ratedmovies.dat",
    sep="\t",
    encoding="latin-1"
)

movies = pd.read_csv(
    DATA_PATH / "movies.dat",
    sep="\t",
    encoding="latin-1"
)

genres = pd.read_csv(
    DATA_PATH / "movie_genres.dat",
    sep="\t",
    encoding="latin-1"
)

tags = pd.read_csv(
    DATA_PATH / "movie_tags.dat",
    sep="\t",
    encoding="latin-1"
)

actors = pd.read_csv(
    DATA_PATH / "movie_actors.dat",
    sep="\t",
    encoding="latin-1"
)

directors = pd.read_csv(
    DATA_PATH / "movie_directors.dat",
    sep="\t",
    encoding="latin-1"
)

countries = pd.read_csv(
    DATA_PATH / "movie_countries.dat",
    sep="\t",
    encoding="latin-1"
)

print("Main dataset files loaded successfully!")
```

```python
tag_names = pd.read_csv(
    DATA_PATH / "tags.dat",
    sep="\t",
    encoding="latin-1"
)

# Rename columns to make the join easier to understand
tag_names = tag_names.rename(columns={
    "id": "tagID",
    "value": "tag"
})

print(tag_names.head())
```

```python
# Compare movie IDs across datasets

movie_ids = set(movies["id"].dropna().unique())
rating_movie_ids = set(ratings["movieID"].dropna().unique())

print("Movies in catalogue:", len(movie_ids))
print("Movies with ratings:", len(rating_movie_ids))

print("Rated movies missing from catalogue:",
      len(rating_movie_ids - movie_ids))

print("Catalogue movies without ratings:",
      len(movie_ids - rating_movie_ids))
```

```python
movies_clean = movies.copy()
genres_clean = genres.copy()
tags_clean = tags.copy()
actors_clean = actors.copy()
directors_clean = directors.copy()
countries_clean = countries.copy()
ratings_clean = ratings.copy()
```

```python
genre_features = (
    genres_clean
    .dropna(subset=["genre"])
    .groupby("movieID")["genre"]
    .apply(lambda x: " ".join(sorted(set(x.astype(str)))))
    .reset_index(name="genres")
)

display(genre_features.head())
```

```python
tag_data = tags_clean.merge(
    tag_names,
    on="tagID",
    how="left"
)

tag_data = tag_data.dropna(subset=["tag"])

# Prioritise tags with higher weights
tag_data = tag_data.sort_values(
    ["movieID", "tagWeight"],
    ascending=[True, False]
)

tag_features = (
    tag_data
    .groupby("movieID")["tag"]
    .apply(lambda x: " ".join(dict.fromkeys(x.astype(str))))
    .reset_index(name="tags")
)

display(tag_features.head())
```

```python
actor_data = actors_clean.dropna(
    subset=["actorName"]
).copy()

actor_data = actor_data.sort_values(
    ["movieID", "ranking"]
)

actor_features = (
    actor_data
    .groupby("movieID")["actorName"]
    .apply(lambda x: " ".join(
        x.drop_duplicates().astype(str).head(10)
    ))
    .reset_index(name="actors")
)

display(actor_features.head())
```

```python
director_features = (
    directors_clean
    .dropna(subset=["directorName"])
    .groupby("movieID")["directorName"]
    .apply(lambda x: " ".join(
        x.drop_duplicates().astype(str)
    ))
    .reset_index(name="directors")
)

display(director_features.head())
```

```python
director_features = (
    directors_clean
    .dropna(subset=["directorName"])
    .groupby("movieID")["directorName"]
    .apply(lambda x: " ".join(
        x.drop_duplicates().astype(str)
    ))
    .reset_index(name="directors")
)

display(director_features.head())
```

```python
country_features = (
    countries_clean
    .dropna(subset=["country"])
    .groupby("movieID")["country"]
    .apply(lambda x: " ".join(
        sorted(set(x.astype(str)))
    ))
    .reset_index(name="countries")
)

display(country_features.head())
```

```python
movie_features = movies_clean.copy()

# Standardise the movie ID column
movie_features = movie_features.rename(
    columns={"id": "movieID"}
)

# Merge all aggregated metadata
feature_tables = [
    genre_features,
    tag_features,
    actor_features,
    director_features,
    country_features
]

for feature_df in feature_tables:
    movie_features = movie_features.merge(
        feature_df,
        on="movieID",
        how="left"
    )

# Fill missing metadata with empty strings
feature_columns = [
    "genres",
    "tags",
    "actors",
    "directors",
    "countries"
]

movie_features[feature_columns] = (
    movie_features[feature_columns].fillna("")
)

print("Combined movie feature shape:", movie_features.shape)

display(
    movie_features[
        ["movieID", "title", "genres", "tags",
         "actors", "directors", "countries"]
    ].head(10)
)
```

```python
movie_features["content"] = (
    movie_features["genres"] + " " +
    movie_features["tags"] + " " +
    movie_features["actors"] + " " +
    movie_features["directors"] + " " +
    movie_features["countries"]
).str.strip()

display(
    movie_features[
        ["movieID", "title", "content"]
    ].head()
)
```

```python
print("Total movies:", len(movie_features))

print("Unique movie IDs:",
      movie_features["movieID"].nunique())

print("Duplicate movie IDs:",
      movie_features["movieID"].duplicated().sum())

print("\nMissing values:")
print(movie_features[
    ["title", "genres", "tags", "actors",
     "directors", "countries", "content"]
].isnull().sum())

print("\nMovies with empty content:",
      (movie_features["content"].str.len() == 0).sum())
```

```python
PROCESSED_PATH = PROJECT_ROOT / "dataset" / "processed"

PROCESSED_PATH.mkdir(
    parents=True,
    exist_ok=True
)

ratings_clean.to_csv(
    PROCESSED_PATH / "ratings_clean.csv",
    index=False
)

movie_features.to_csv(
    PROCESSED_PATH / "movie_features.csv",
    index=False
)

print("Processed datasets saved successfully!")
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_parts = []
test_parts = []

for user_id, user_data in ratings_clean.groupby("userID"):

    user_train, user_test = train_test_split(
        user_data,
        test_size=0.20,
        random_state=42
    )

    train_parts.append(user_train)
    test_parts.append(user_test)

train_ratings = pd.concat(train_parts).reset_index(drop=True)
test_ratings = pd.concat(test_parts).reset_index(drop=True)

print("Training interactions:", len(train_ratings))
print("Testing interactions:", len(test_ratings))
```

```python
train_users = set(train_ratings["userID"].unique())
train_movies = set(train_ratings["movieID"].unique())

test_ratings = test_ratings[
    test_ratings["userID"].isin(train_users) &
    test_ratings["movieID"].isin(train_movies)
].reset_index(drop=True)

print("Final training interactions:", len(train_ratings))
print("Final testing interactions:", len(test_ratings))

print(
    "Users in training:",
    train_ratings["userID"].nunique()
)

print(
    "Movies in training:",
    train_ratings["movieID"].nunique()
)
```

```python
assert len(train_ratings) > 0
assert len(test_ratings) > 0

assert set(test_ratings["userID"]).issubset(
    set(train_ratings["userID"])
)

assert set(test_ratings["movieID"]).issubset(
    set(train_ratings["movieID"])
)

print("Warm-start split validation successful!")
```
